```python
API_BASE_URL = "https://ai-prediksi-kelulusan.vercel.app"

print("API:", API_BASE_URL)
```

```python
import requests

url = f"{API_BASE_URL}/v2/prediksi/80"

try:
    response = requests.get(url, timeout=20)

    print("Status:", response.status_code)

    response.raise_for_status()

    data = response.json()

    print("Hasil prediksi:", data)

except requests.exceptions.Timeout:
    print("Error: API terlalu lama merespons")

except requests.exceptions.RequestException as error:
    print("Error koneksi atau HTTP:", error)
```

```python
daftar_nilai = [60, 75, 90]

for nilai in daftar_nilai:
    url = f"{API_BASE_URL}/v2/prediksi/{nilai}"

    try:
        response = requests.get(url, timeout=20)
        response.raise_for_status()

        data = response.json()

        print(
            "Nilai:", data["nilai"],
            "| Prediksi:", data["prediksi"],
            "| Versi:", data["version"]
        )

    except requests.exceptions.RequestException as error:
        print("Gagal menguji nilai", nilai, ":", error)
```
